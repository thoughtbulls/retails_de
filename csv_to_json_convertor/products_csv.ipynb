retails db data convert from csv to json

```python
schema = """
    product_id INTEGER,
    product_category_id INTEGER,
    product_name STRING,
    product_description STRING,
    product_price DOUBLE,
    product_image STRING
    """

df = spark.read \
        .format("csv") \
        .schema(schema) \
        .option("inferSchema", "true") \
        .load("dbfs:/Volumes/workspace/retails/raw_data/csv/products/")
```

```python
df.printSchema()
```

```python
display(df)
```

```python
df.write \
.format("json") \
.mode("overwrite") \
.save("dbfs:/Volumes/workspace/retails/raw_data/json/products/")
```
