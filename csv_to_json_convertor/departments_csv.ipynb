retails db data convert from csv to json

```python
schema = """
    department_id INT,
    department_name STRING
    """

df = spark.read \
        .format("csv") \
        .schema(schema) \
        .option("inferSchema", "true") \
        .load("dbfs:/Volumes/workspace/retails/raw_data/csv/departments/")
```

```python
df.printSchema()
```

```python
display(df)
```

```python
df.write \
.format("json") \
.mode("overwrite") \
.save("dbfs:/Volumes/workspace/retails/raw_data/json/departments/")
```
