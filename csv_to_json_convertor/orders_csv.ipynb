retails db data convert from csv to json

```python
schema = """
    order_id INTEGER,
    order_date STRING,
    order_customer_id INTEGER,
    order_status STRING
    """

df = spark.read \
        .format("csv") \
        .schema(schema) \
        .option("inferSchema", "true") \
        .load("dbfs:/Volumes/workspace/retails/raw_data/csv/orders/")
```

```python
df.printSchema()
```

```python
df.write \
.format("json") \
.mode("overwrite") \
.save("dbfs:/Volumes/workspace/retails/raw_data/json/orders/")
```
