retails db data convert from csv to json

```python
schema = """
    order_items INTEGER,
    order_item_order_id INTEGER,
    order_item_product_id INTEGER,
    order_item_quantity INTEGER,
    order_item_subtotal FLOAT,
    order_item_product_price FLOAT
    """

df = spark.read \
        .format("csv") \
        .schema(schema) \
        .option("inferSchema", "true") \
        .load("dbfs:/Volumes/workspace/retails/raw_data/csv/order_items/")
```

```python
df.printSchema()
```

```python
df.write \
.format("json") \
.mode("overwrite") \
.save("dbfs:/Volumes/workspace/retails/raw_data/json/order_items/")
```
