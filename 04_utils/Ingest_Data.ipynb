```python
#  Now ingest data using COPY-INTO
orders_df = spark.sql("""
COPY INTO sales.bronze.orders_raw
FROM '/Volumes/data/raw/sales/read_input/orders'

FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header' = 'true',
  'inferSchema' = 'false'
)
COPY_OPTIONS (
    'mergeSchema' = 'true'
)
""")

```
