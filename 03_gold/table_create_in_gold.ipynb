```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.dim_products (

    product_key BIGINT GENERATED ALWAYS AS IDENTITY,

    product_id BIGINT,
    product_name STRING,
    product_description STRING,
    product_price DOUBLE,
    product_image STRING,
    product_category_id BIGINT,

    is_active BOOLEAN,

    effective_from TIMESTAMP,
    effective_to TIMESTAMP,
    is_current BOOLEAN,

    created_ts TIMESTAMP,
    updated_ts TIMESTAMP
)
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.dim_departments (

    department_key BIGINT GENERATED ALWAYS AS IDENTITY,

    department_id INT,
    department_name STRING,

    is_active BOOLEAN,

    is_current BOOLEAN,
    effective_from TIMESTAMP,
    effective_to TIMESTAMP,

    created_ts TIMESTAMP,
    updated_ts TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.dim_categories (

    category_key BIGINT GENERATED ALWAYS AS IDENTITY,

    category_id INT,
    category_department_id INT,
    category_name STRING,

    is_active BOOLEAN,

    is_current BOOLEAN,
    effective_from TIMESTAMP,
    effective_to TIMESTAMP,

    created_ts TIMESTAMP,
    updated_ts TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.dim_customers (

    customer_key BIGINT GENERATED ALWAYS AS IDENTITY,

    customer_id INT,

    customer_fname STRING,
    customer_lname STRING,
    customer_email STRING,

    customer_street STRING,
    customer_city STRING,
    customer_state STRING,
    customer_zipcode INT,

    is_active BOOLEAN,
    
    is_current BOOLEAN,
    effective_from TIMESTAMP,
    effective_to TIMESTAMP,

    created_ts TIMESTAMP,
    updated_ts TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.dim_dates (

    date_key INT,

    full_date DATE,
    year INT,
    quarter INT,
    month INT,
    month_name STRING,
    week_of_year INT,
    day_of_month INT,
    day_of_week INT,
    day_name STRING,
    is_weekend BOOLEAN
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.fact_orders (

    order_fact_key BIGINT GENERATED ALWAYS AS IDENTITY,

    order_id BIGINT,

    customer_key BIGINT,
    order_date_key INT,

    order_status STRING,

    total_order_amount DECIMAL(10, 2),
    total_items BIGINT,

    created_ts TIMESTAMP
)
USING DELTA
PARTITIONED BY (order_date_key)
```

```python
%sql
CREATE TABLE IF NOT EXISTS retails.gold.fact_order_items (

    order_item_fact_key BIGINT GENERATED ALWAYS AS IDENTITY,

    order_item_id BIGINT,
    order_id BIGINT,

    customer_key BIGINT,
    product_key BIGINT,
    category_key BIGINT,
    department_key BIGINT,

    order_date_key BIGINT,

    quantity INT,
    product_price DOUBLE,
    subtotal DOUBLE,

    created_ts TIMESTAMP
);
```
