Generally we use ingestion system like:
1. (fixed schema + cloudFiles.schemaEvolutionMode = rescue) + (_rescued_data + _corrupt_record) == ERP, SAS

- Why?
    Because:
    - source contracts usually exist
    - structure mostly stable
    - but unexpected fields still happen
    - pipelines should remain resilient

- Typical systems
        ✔ ERP
        ✔ Salesforce
        ✔ SAP
        ✔ SaaS APIs
        ✔ CDC feeds
2. (strict schema + cloudFiles.schemaEvolutionMode = failOnNewColumns) + (_rescued_data + _corrupt_record) == Banking, follow the compliance.

- Why?
    Because:
    - unexpected data may indicate risk
    - audit/compliance requirements
    - downstream calculations sensitive

- Typical systems
        ✔ Banking core system
        ✔ Payment System
        ✔ Trading Capital market
        ✔ Insurance systems
        ✔ Healthcare systems
        ✔ Govermant/Tax tax

3. (no schema + cloudFiles.schemaEvolutionMode = addNewColumns + schemaHints) + (_rescued_data + _corrupt_record) == logs , event streams, telemetry, etc

- Why?
    Because:
    - no predefined schema
    - dynamic JSON payloads
    - frequent evolution

- Typical systems
        ✔ IoT
        ✔ clickstream
        ✔ application logs
        ✔ Kafka events
        ✔ telemetry

```python
schema = """
        department_id STRING,
        department_name STRING
        """

# ingest departments
df = (spark.readStream \
    .format("cloudFiles") \
    .option("cloudFiles.format", "json") 
    .option("cloudFiles.schemaLocation", "dbfs:/Volumes/data/raw/_schemas/retails/dev/departments/") \
    .option("cloudFiles.maxFilesPerTrigger", "1") \
    .schema(schema) \
    .option("cloudFiles.schemaEvolutionMode", "rescue") \
    .option("cloudFiles.rescuedDataColumn", "_rescued_data") \
    .option("cloudFiles.schemaHints", "department_id INTEGER, _corrupt_record STRING") \
    .load("dbfs:/Volumes/data/raw/dev/retail_db/departments/")
)
```

```python
from pyspark.sql.functions import to_json, struct, col, current_timestamp, current_date, regexp_extract, expr, lit, StringType, coalesce, when

df = df.withColumn("raw_data", to_json(struct("*"))) \
        .withColumn(
            "source_file_name",
            regexp_extract(col("_metadata.file_path"), r'([^/]+$)', 1)
        ) \
        .withColumn("source_file_path", col("_metadata.file_path")) \
        .withColumn("replay_flag", lit(False))


if "op" in df.columns:
    df = df.withColumn("op", col("op").cast(StringType()))
else:
    df = df.withColumn("op", lit("INSERT").cast(StringType()))

```

```python
from pyspark.sql.functions import *
from datetime import datetime

run_id = datetime.now().strftime("%Y%m%d_%H%M%S")

def process_batch(df, batch_id):

    df = (
        df
        .withColumn("department_id", col("department_id").cast("String")) \
        .withColumn("ingestion_ts", current_timestamp()) \
        .withColumn("ingestion_dt", current_date()) \
        .withColumn("batch_id", lit(batch_id)) \
        .withColumn("run_id", lit(run_id)) \
        .withColumn("source_system", lit("ERP"))
    )

    (
        df.write
            .format("delta") \
            .mode("append") \
            .option("mergeSchema", "true") \
            .saveAsTable("retails.bronze.departments_raw")
    )
```

```python

(df.writeStream \
    .foreachBatch(process_batch) \
    .option("checkpointLocation", "dbfs:/Volumes/data/raw/_checkpoints/retails/dev/bronze/departments/") \
    .trigger(availableNow=True)
    .start()
)
```
